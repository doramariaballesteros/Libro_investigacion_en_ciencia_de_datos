# **1. Lectura y previsualización del dataset**

```python
import pandas as pd
clima_df = pd.read_csv("/content/clima_eda.csv")
clima_df.head()
```

**CONVERTIR LA COLUMNDA DE TIEMPO A ÍNDICE**

```python
clima_df["tiempo_hora_local"] = pd.to_datetime(clima_df["tiempo_hora_local"])
clima_df = clima_df.set_index("tiempo_hora_local")
```

```python
clima_df.head()
```

# **2.  Feature Engineering**

**ROLLING**

La técnica **Rolling** permite construir nuevas características a partir de una **ventana móvil de observaciones anteriores**.

Para cada instante de tiempo, se seleccionan los últimos \(n\) valores de una variable y se calcula una medida estadística (ej. promedio). A medida que avanza el tiempo, la ventana también se desplaza.

Por ejemplo, un **Rolling de 24 muestras** representa el comportamiento de una variable durante las últimas 24 horas, siempre y cuando el espaciamiento entre muestras sea de 1 hora.

**Nota aclaratoria:** se utiliza `shift(1)` antes de aplicar Rolling para garantizar que la característica se construya utilizando **únicamente información pasada**, sin incluir el valor del instante que se desea predecir.

Se evaluarán diferentes **variables climáticas** y **tamaños de ventana** para analizar qué información temporal aporta mayor capacidad predictiva al modelo.

```python
columnas_rolling = [
    "temperatura",
    "humedad",
    "presion",
    "punto_rocio",
    "direccion_viento",
    "velocidad_viento",
    "rafagas_viento",
    "presion",
    "indice_UV",
    "ozono",
    "intensidad_precipitacion"
]

ventanas = [12, 24, 48, 72]
```

```python
for columna in columnas_rolling:
    for ventana in ventanas:
        clima_df[f"{columna}_rolling_{ventana}"] = (
            clima_df[columna]
            .shift(1)
            .rolling(window=ventana)
            .mean()
        )
```

```python
clima_df.head(20)
```

```python
clima_df = clima_df.dropna()
```

# **3.  MODELAMIENTO**

```python
# Variable objetivo
y = clima_df['estado_clima_num']

# Variables predictoras
X = clima_df.drop(columns=['estado_clima_num'])

print("Dimensión de X:", X.shape)
print("Dimensión de y:", y.shape)
```

```python
# Punto de corte: 70 % entrenamiento - 30 % prueba
punto_corte = int(len(clima_df) * 0.7)

# División temporal
X_train = X.iloc[:punto_corte]
X_test  = X.iloc[punto_corte:]

y_train = y.iloc[:punto_corte]
y_test  = y.iloc[punto_corte:]

print("Observaciones de entrenamiento:", len(X_train))
print("Observaciones de prueba:", len(X_test))

print("\nPeriodo de entrenamiento:")
print(X_train.index.min(), "a", X_train.index.max())

print("\nPeriodo de prueba:")
print(X_test.index.min(), "a", X_test.index.max())
```

```python
from sklearn.ensemble import RandomForestClassifier

# Modelo FE - Rolling
modelo = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Entrenamiento
modelo.fit(X_train, y_train)
```

```python
# Predicciones
y_pred = modelo.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report

# Accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", round(accuracy, 4))
print()

# Reporte por clase
print(classification_report(
    y_test,
    y_pred,
    zero_division=0
))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=[1, 2, 3, 4, 5]
)

fig, ax = plt.subplots(figsize=(6, 6))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=[
        'clear-day',
        'clear-night',
        'partly-cloudy-day',
        'partly-cloudy-night',
        'cloudy'
    ]
)

disp.plot(ax=ax, xticks_rotation=45)

plt.title('Matriz de confusión - Modelo FE - Rolling')
plt.tight_layout()
plt.show()
```

### Importancia de las características

Después de aplicar **Feature Engineering**, podemos analizar cuáles de las características tienen mayor influencia en las decisiones del modelo.

En un Random Forest, la importancia de una característica indica cuánto contribuye a reducir la impureza en los árboles que conforman el modelo. Un valor mayor indica que esa característica fue utilizada con mayor relevancia durante el proceso de clasificación.

Este análisis nos permitirá observar si las nuevas características construidas a partir del comportamiento temporal de los datos están aportando información útil al modelo.

```python
# Importancia de las características

import pandas as pd
import matplotlib.pyplot as plt

importancias = pd.Series(
    modelo.feature_importances_,
    index=X_train.columns
).sort_values(ascending=True)

plt.figure(figsize=(14, 8))
importancias.plot(kind="barh")

plt.xlabel("Importancia")
plt.ylabel("Característica")
plt.title("Importancia de las características FE - Rolling")

plt.tight_layout()
plt.show()
```

### 🔎 Análisis de los nuevos features

Para generar los nuevos *features* se utilizaron ventanas de **12, 24, 48 y 72 horas**. En todos los casos se aplicó previamente `shift(1)`, de manera que el cálculo en el instante $t$ utiliza únicamente valores anteriores.

> ⚠️ **Recuerda:** el uso de `shift(1)` evita *data leakage*, ya que el valor actual no se utiliza para construir el *feature* con el que posteriormente se realizará la predicción.

Al analizar la importancia de las características se observa que algunos de los nuevos *features* generados mediante **Rolling** tienen incluso mayor importancia que la variable original.

Pero hay otro resultado interesante: **el mejor tamaño de ventana no es necesariamente el mismo para todos los features**.
