```python

import ipywidgets as widgets
from IPython.display import display, Markdown
```

```python

import ipywidgets as widgets
from IPython.display import display, Markdown

# Diccionario para almacenar las respuestas
respuestas = {}

def mostrar_pregunta(numero, enunciado, opciones):
    display(Markdown(f"### 📝 Pregunta {numero}\n**{enunciado}**"))

    selector = widgets.RadioButtons(
        options=[(f"{letra}. {texto}", letra)
                 for letra, texto in zip("ABCD", opciones)],
        value=None,
        layout=widgets.Layout(width="100%"),
        style={"description_width": "initial"}
    )

    def guardar_respuesta(change):
        respuestas[numero] = change["new"]

    selector.observe(guardar_respuesta, names="value")
    display(selector)
    return selector
```

```python

pregunta_1 = mostrar_pregunta(
    1,
    "¿Cuál es la hipótesis principal de la investigación?",
    [
        "Las señales de voz pueden adaptarse al modificar su frecuencia de muestreo.",
        "Las señales de voz pueden adaptarse al reorganizar sus coeficientes wavelet.",
        "Las señales de voz pueden adaptarse al eliminar sus componentes de ruido.",
        "Las señales de voz pueden adaptarse al igualar sus amplitudes temporales."
    ]
)
```

```python

pregunta_2 = mostrar_pregunta(
    2,
    "¿Qué función cumple la Transformada Wavelet Discreta (DWT) en el procedimiento?",
    [
        "Obtener coeficientes de aproximación y detalle de las señales.",
        "Obtener coeficientes de correlación y covarianza de las señales.",
        "Obtener componentes de amplitud y fase de las señales.",
        "Obtener componentes de frecuencia y energía de las señales."
    ]
)
```

```python

pregunta_3 = mostrar_pregunta(
    3,
    "¿Cuál de las siguientes condiciones es necesaria para la adaptación?",
    [
        "Las señales deben tener la misma amplitud y el mismo idioma.",
        "Las señales deben tener la misma duración y el mismo hablante.",
        "Las señales deben tener la misma energía y el mismo contenido.",
        "Las señales deben tener el mismo marco temporal y la misma base wavelet."
    ]
)
```

```python

pregunta_4 = mostrar_pregunta(
    4,
    "¿Qué mide el coeficiente de correlación de Pearson al cuadrado (r²) en los experimentos?",
    [
        "La diferencia de ruido entre las señales comparadas.",
        "La diferencia de duración entre las señales comparadas.",
        "La similitud entre los coeficientes wavelet de las señales comparadas.",
        "La similitud entre la energía de las señales comparadas."
    ]
)
```
